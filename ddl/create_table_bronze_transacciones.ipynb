```sql
CREATE OR REPLACE TABLE bronze.transacciones (
    fecha TIMESTAMP COMMENT 'Fecha y hora de la transacción en UTC — ISO 8601',
    tipoTran STRING COMMENT 'Tipo de transacción: Compra/Venta',
    id_cliente STRING COMMENT 'Identificador único ofuscado. Formato:CLI + 8 hex',
    descripcion_titulo STRING COMMENT 'Nombre completo del instrumento',
    moneda STRING COMMENT 'ARS o USD. Sufijo D en símbolo = dólarcable',
    simbolo_titulo STRING COMMENT 'Ticker: AL30, AL30D, MSFT, YPFD, GLD…',
    cantidad INT COMMENT 'Nominales / láminas operada',
    precio DOUBLE COMMENT 'Precio unitario de ejecución',
    id_transaccion STRING COMMENT 'ID único ofuscado. Formato: TXN + 13alfanum',
    origen STRING COMMENT 'Origen de la transacción: App Mobile · Web Desktop · Web Responsive · API · IOLnet',
    fecha_particion DATE COMMENT 'Fecha de partición para optimización a partir de la fecha de la transacción',
    fecha_auditoria TIMESTAMP COMMENT 'Fecha y hora de auditoría'
)
USING DELTA
PARTITIONED BY (fecha_particion)
COMMENT 'Tabla que almacena las transacciones realizadas por clientes en IOL, incluyendo detalles de títulos, precios, cantidades y auditoría.'
```
