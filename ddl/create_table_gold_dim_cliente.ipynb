```sql
CREATE OR REPLACE TABLE gold.dim_cliente (
    cliente_sk BIGINT GENERATED ALWAYS AS IDENTITY COMMENT 'Surrogate key (auto-incremental)',
    id_cliente STRING COMMENT 'Identificador único del cliente (natural key)',
    dias_activos INT COMMENT 'Días con al menos una operación',
    total_transacciones BIGINT COMMENT 'Total de transacciones del cliente',
    volumen_total DECIMAL(18,2) COMMENT 'Volumen total operado (suma de importes)',
    primera_transaccion TIMESTAMP COMMENT 'Fecha de primera transacción',
    ultima_transaccion TIMESTAMP COMMENT 'Fecha de última transacción',
    CONSTRAINT pk_dim_cliente PRIMARY KEY (cliente_sk),
    CONSTRAINT uk_dim_cliente_id UNIQUE (id_cliente)
)
USING DELTA
COMMENT 'Dimensión de clientes con métricas agregadas de actividad';
```
