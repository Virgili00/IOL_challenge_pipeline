```sql
CREATE OR REPLACE TABLE gold.dim_instrumento (
    instrumento_sk BIGINT GENERATED ALWAYS AS IDENTITY COMMENT 'Surrogate key (auto-incremental)',
    simbolo_base STRING COMMENT 'Símbolo normalizado sin sufijos (ej: AL30, AAPL)',
    simbolo_original STRING COMMENT 'Símbolo original con variantes (ej: AL30D, YPFD)',
    tipo_instrumento STRING COMMENT 'Tipo: ACCION, BONO, CEDEAR, ON, ETF, etc',
    variante_liquidacion STRING COMMENT 'STANDARD (pesos), MEP (dólar cable)',
    descripcion_titulo STRING COMMENT 'Nombre completo del instrumento',
    CONSTRAINT pk_dim_instrumento PRIMARY KEY (instrumento_sk),
    CONSTRAINT uk_dim_instrumento UNIQUE (simbolo_base, variante_liquidacion)
)
USING DELTA
COMMENT 'Dimensión de instrumentos financieros con símbolos y variantes de liquidación';
```
