```sql
CREATE OR REPLACE TABLE gold.fact_transacciones (
    id_transaccion STRING COMMENT 'ID único de transacción (dimensión degenerada)',
    fecha_sk BIGINT COMMENT 'FK a dim_fecha (surrogate key)',
    instrumento_sk BIGINT COMMENT 'FK a dim_instrumento (surrogate key)',
    cliente_sk BIGINT COMMENT 'FK a dim_cliente (surrogate key)',
    origen_sk BIGINT COMMENT 'FK a dim_origen (surrogate key)',
    tipo_transaccion STRING COMMENT 'Compra / Venta',
    cantidad INT COMMENT 'Cantidad operada (nominales/láminas)',
    precio_transaccion DECIMAL(18,4) COMMENT 'Precio unitario de ejecución',
    precio_mercado DECIMAL(18,4) COMMENT 'Precio consolidado: API o fallback a precio transacción',
    importe DECIMAL(18,2) COMMENT 'Importe total = cantidad × precio_mercado_final',
    moneda STRING COMMENT 'Moneda: ARS o USD',
    fuente_precio STRING COMMENT 'Fuente del precio: API (Yahoo Finance) o TRANSACTION_PRICE',
    fecha_auditoria TIMESTAMP COMMENT 'Timestamp de procesamiento',
    CONSTRAINT pk_fact_transacciones PRIMARY KEY (id_transaccion)
)
USING DELTA
COMMENT 'Fact table de transacciones con precios de mercado enriquecidos mediante fallback automático.';
```
