```sql
CREATE OR REPLACE TABLE gold.dim_origen (
    origen_sk BIGINT GENERATED ALWAYS AS IDENTITY COMMENT 'Surrogate key (auto-incremental)',
    origen STRING COMMENT 'Canal de operación (natural key): App Mobile, Web Desktop, API, IOLnet',
    CONSTRAINT pk_dim_origen PRIMARY KEY (origen_sk),
    CONSTRAINT uk_dim_origen UNIQUE (origen)
)
USING DELTA
COMMENT 'Dimensión de canales de transacción';
```
