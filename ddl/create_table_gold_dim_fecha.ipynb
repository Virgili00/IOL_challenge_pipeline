```sql
CREATE OR REPLACE TABLE gold.dim_fecha (
    fecha_sk BIGINT GENERATED ALWAYS AS IDENTITY COMMENT 'Surrogate key (auto-incremental)',
    fecha DATE COMMENT 'Fecha completa (natural key)',
    anio INT COMMENT 'Año',
    mes INT COMMENT 'Mes (1-12)',
    trimestre INT COMMENT 'Trimestre (1-4)',
    dia INT COMMENT 'Día del mes (1-31)',
    dia_semana INT COMMENT 'Día de la semana (1=Domingo, 7=Sábado)',
    semana_anio INT COMMENT 'Semana del año (1-53)',
    es_dia_habil BOOLEAN COMMENT 'Indica si es día hábil (lun-vie)',
    nombre_mes STRING COMMENT 'Nombre del mes',
    nombre_dia STRING COMMENT 'Nombre del día',
    CONSTRAINT pk_dim_fecha PRIMARY KEY (fecha_sk),
    CONSTRAINT uk_dim_fecha_fecha UNIQUE (fecha)
)
USING DELTA
COMMENT 'Dimensión calendario con atributos temporales';
```
