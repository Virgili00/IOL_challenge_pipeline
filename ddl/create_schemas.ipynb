```sql
-- Schemas de la arquitectura Medallion
CREATE SCHEMA IF NOT EXISTS bronze 
  COMMENT 'Raw data - ingesta de archivos CSV y APIs externas';

CREATE SCHEMA IF NOT EXISTS silver 
  COMMENT 'Cleaned data - normalización, validación y enriquecimiento';

CREATE SCHEMA IF NOT EXISTS gold 
  COMMENT 'Business-level aggregates - star schema para análisis';

CREATE SCHEMA IF NOT EXISTS refined 
  COMMENT 'Alias de gold - star schema para reporting';
```
