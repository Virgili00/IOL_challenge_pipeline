```sql

CREATE TABLE IF NOT EXISTS audit.pipeline_errors (
    error_id STRING,
    process_datetime TIMESTAMP,
    runId STRING,
    layer STRING,
    table_name STRING,
    notebook_path STRING,
    error_message STRING,
    error_detail STRING,
    execution_timestamp TIMESTAMP
)
USING DELTA;

CREATE TABLE IF NOT EXISTS audit.test_failed_records (
        test_id STRING,
        run_id STRING,
        table_name STRING,
        test_name STRING,
        record_id STRING,
        failure_reason STRING,
        created_at TIMESTAMP
    )
USING DELTA;
```
